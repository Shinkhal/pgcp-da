```python
import numpy as np
import scipy.stats as stats

np.random.seed(43)
clean_data = np.random.normal(scale=10, loc=50, size=1000)
# print(clean_data)

corrupted_data = clean_data.copy()

corrupted_data[0] = 1e8
clean_mean = np.mean(clean_data)
clean_median = np.median(clean_data)
corrupted_mean = np.mean(corrupted_data)
corrupted_median = np.median(corrupted_data)

print(f"Clean Data - Mean: {clean_mean}, Median: {clean_median}")
print(f"Corrupted Data - Mean: {corrupted_mean}, Median: {corrupted_median}")
```

### Whats is Dispersion ?

- measure of spreads, variability or scale of the data around its central location
- Dispersion referes to the extent to which a distribution



```python
population = np.random.normal(loc=100.0, scale=25.0, size=10000)
# population = [20,25,15,30]
print(population)
true_var = np.var(population)
print(true_var)

true_std = np.sqrt(true_var)
print(true_std)

sample_var = np.var(population, ddof=1)
print(sample_var)

```

### Generate Fake age data for 500 people

```python
ages = np.random.randint(low=18, high=90, size=500)
print("Ages: ", ages)
```

```python
from scipy import stats

print("modes of ages : ", stats.mode(ages))
```

```python
from collections import Counter
age_counts = Counter(ages)
age_clean = {int(k): v for k,v in age_counts.items()}
print(age_clean)
```

# What is Standar Deviation ?

```python
from scipy.stats import norm

import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-3, 3, 0.001)
y = norm.pdf(x)
plt.plot(x, y)
plt.show()
```
