## Load the Dataset

```python
import pandas as pd

df = pd.read_csv("../ASSETS/drinks.csv")
print(df.head())
```

# Initial Inspection to get Shape

```python
print("total rows (Countries) : ", df.shape[0])
print("total columns (Features) : ", df.shape[1])
```

### Schema Health Checking
#### The Goal : Ensure every column has correct data type

```python
print(df.info())
```

### Spot Missing Values & Imputation
- The Goal : Ensure every column has correct data type
- Idea : Check for missing values and handle them appropriately (e.g., imputation, removal)

```python
missing_values = df.isnull().sum()
print("Missing Values : \n", missing_values)
```

#### Data Sanitization

```python
df['continent'] = df['continent'].fillna('Unknown')
```

```python
# print(df[df['country'] == 'Sweden'])
# print(df.query('country == "Sweden"')[['country', 'continent']])
# print(df.loc[df['country'] == 'Sweden', ['country', 'continent']])

print(df[df['country'].isin(['Sweden','India'])][['country', 'continent']])
```

```python
print(df[["beer_servings", "wine_servings", "spirit_servings"]].describe().T[["mean", "std", "min", "25%", "50%", "75%", "max"]])
```

```python
beer_serving_negative_count = (df['beer_servings'] < 0).sum()
print("Negative Beer Servings Count: ", beer_serving_negative_count)
```

### Lets fix Negatives
- either feed zero or filter out countries with negative values

```python
df_copy = df.copy()
print("df dimensions before filtering negative beer servings: ", df.shape)
df = df[df['beer_servings'] >= 0]
print("df dimensions after filtering negative beer servings: ", df.shape)
```

### String Sanitization


```python
# get all the features name in a list
features_name = df.columns.to_list()
print(features_name)

# assume identifying leading or trailing spaces in feature name - continent
df['continent'] = df['continent'].str.strip().str.title()
print(df['continent'].head())
```

```python
# print(df['continent'].head())
print(df['continent'].unique())
```

## Grouping & Aggregation

```python
average_beer_by_continent = df.groupby('continent')['beer_servings'].mean().round(2)
print("Average Beer Servings by Continent:\n", average_beer_by_continent)
```

```python
top_drinker = df.loc[df['total_litres_of_pure_alcohol'].idxmax()]
print(top_drinker)
```
