```python
import pandas as pd
from scipy.stats import zscore, chi2_contingency

df = pd.DataFrame({
    "employee": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "salary": [30000, 32000, 31000, 35000, 33000,
               34000, 36000, 37000, 39000, 180000],
    "experience": [1, 2, 2, 3, 3, 4, 5, 5, 6, 8]
})
```

```python
q1 = df['salary'].quantile(0.25)
q3 = df['salary'].quantile(0.75)

IQR = q3 - q1

lower_bound = q1 - 1.5 * IQR
upper_bound = q3 + 1.5 * IQR


print(IQR)
print(lower_bound, upper_bound)
```

```python
outliers = df[(df['salary']< lower_bound) | (df['salary']> upper_bound)]

print(outliers)

# outlier using zscore
df['z_score'] = zscore(df['salary'])
outliers_zscore = df[df['z_score'].abs() > 3]

print(outliers_zscore)
```

```python
df['salary'] = df['salary'].clip(lower=lower_bound, upper=upper_bound)
print(df)
```

# Chi Square Test


```python

df_chi = pd.DataFrame({
    "course": ["Python", "Python", "Python", "Python",
               "Java", "Java", "Java", "Java",
               "SQL", "SQL", "SQL", "SQL"],
    "outcome": ["Pass", "Pass", "Fail", "Pass",
                "Pass", "Fail", "Fail", "Pass",
                "Pass", "Pass", "Pass", "Fail"]
})
```

```python
table = pd.crosstab(df_chi['course'], df_chi['outcome'])
print(table)

chi2, p_value, dof, expected = chi2_contingency(table)

print('_' * 50)
print(chi2)
print(p_value)
print(dof)
print(expected)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
