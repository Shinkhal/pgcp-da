```python
from xml.etree.ElementInclude import include

import pandas as pd

df = pd.DataFrame({
    "Student Name": [" aman ", "RIYA", "karan", "uTSAV", "  RiYa"],
    "Marks Obtained": ["35", "72", "72", "Fail", "85"]
})
```

```python
#snake case in columns

df.columns = (df.columns.str.strip().str.lower().str.replace(' ','_'))

print(df)
```

```python
# sanitize text

string_cols = df.select_dtypes(include=["object","string"]).columns
df[string_cols] = df[string_cols].apply(lambda col : col.str.strip().str.lower())


if 'student_name' in df.columns:
    df['student_name'] = df['student_name'].str.title()

print(df)
```

```python
# convert marks from str to int

# df['marks_obtained']= df['marks_obtained'].astype('Int64')

df['marks_obtained'] = pd.to_numeric(df["marks_obtained"], errors='coerce')
print(df.dtypes)
print(df)
```

```python
# fill missing values
df['marks_obtained'] = df['marks_obtained'].fillna(df['marks_obtained'].median())
print(df)
```

```python
df['Result'] = df['marks_obtained'].apply(
    lambda d : "Pass" if d >= 40 else "Fail"
)

print(df)
```

```python
def grade(marks):
    if marks >=90:
        return 'A+'
    elif marks >=80:
        return 'A'
    elif marks >= 70:
        return 'B+'
    elif marks <= 40:
        return 'F'
    return None


df['Grade'] = df['marks_obtained'].apply(grade)
print(df)
```

```python
# Drop duplicates based only on specific columns
df = df.drop_duplicates(subset=['marks_obtained'])
print(df)
df = df.drop_duplicates(subset=['student_name'])
print(df)
```
