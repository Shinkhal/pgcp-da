# Covariance and Correlation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df = pd.DataFrame({
    "experience": [1, 2, 3, 4, 5, 6, 7, 8],
    "salary": [25, 30, 35, 40, 48, 52, 60, 90],
    "work_hours": [55, 52, 50, 48, 45, 44, 42, 40],
    "satisfaction": [4, 5, 5, 6, 7, 7, 8, 9]
})

# Covariance between exp and salary
print(df["experience"].cov(df["salary"]))
print('-' * 50)
# Pearson correlation between study hours and marks
print(df["experience"].corr(df["salary"]))
print('-' * 50)
# Correlation matrix for all numerical columns
print(df.corr(numeric_only=True))
print('-' * 50)
# Covariance matrix
print(df.cov(numeric_only=True))
```

```python
plt.figure(figsize=(7, 4))
plt.grid(True)
plt.scatter(df["work_hours"], df["salary"])
plt.xlabel("Work Hours")
plt.ylabel("Salary")
plt.title("Work Hours vs Salary")


plt.tight_layout()
plt.show()
```
