# calculating mean, median, mode

```python
import numpy as np
import pandas as pd
from scipy import stats
from scipy.stats import trim_mean, zscore, skew, kurtosis


salary = np.array([25000, 28000, 28000, 30000, 35000, 180000])
```

```python
print(np.mean(salary))
print(np.median(salary))
print(pd.Series(salary).mean())
print(pd.Series(salary).median())
```

```python
print(pd.Series(salary).mode())
print('-'*60)
print(stats.mode(salary, keepdims=False))
```

```python
# trimmed mean
print(trim_mean(salary, 0.20))
```

# Quartiles

```python
q1 = np.percentile(salary, 25)
q2 = np.percentile(salary, 50)
q3 = np.percentile(salary, 75)

print(q1)
print(q3)

IQR = q3 - q1
print(IQR)
print('-'*60)


#quartiles using pd
print(pd.Series(salary).quantile(0.25))
print(pd.Series(salary).quantile(0.75))
print(f'IQR : { (pd.Series(salary).quantile(0.75) - pd.Series(salary).quantile(0.25))}')
```

# Population Variance

```python
print("Population Variance ", np.var(salary))
print("Sample Variance ", np.var(salary, ddof=1))
print("Standard Deviation ",np.std(salary))
```

# Mean absolute deviation (MAD)
### Mean absolute deviation measures the average absolute distance from the mean.

```python
mean = np.mean(salary)
mad = np.mean(np.abs(salary - mean))

print(mad)
```

# Z score
A z-score measures how many standard deviations an observation lies above or below the mean.

```python
print(zscore(salary))
```

# Skewness & Kurtosis

```python
print(skew(salary))
print(kurtosis(salary))
print(kurtosis(salary, fisher=False))
```
