# Skewness

- Skewness is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean.
- It is also known as the curtosis.

### You have startup with 5 employees
salaries = [40,50,45,55,300]

- step 1 - calculate the mean of the salaries
- step 2 - calculate the median of the salaries
- step 3 - calculate the skewness of the salaries
    - decissions
        - if mean > median => positive skewness : average pulled upwards : the tail is long on the right side : its called right skewed
        - if mean < median => negative skewness : average pulled downwards : the tail is long on the left side : its called left skewed
        - if mean = median => zero skewness : average is in the middle : its called neutral : its called symmetric


```python
import matplotlib.pyplot as plt
import numpy as np

salaries = np.array([40, 50, 45, 55, 300])
plt.boxplot(salaries,vert=False)
plt.show()
```

```python
retirement_age = [35 , 62,65,67,71]

mu=np.mean(retirement_age)
median=np.median(retirement_age)
print(mu,median)

skewness=mu-median
print(skewness)
```
