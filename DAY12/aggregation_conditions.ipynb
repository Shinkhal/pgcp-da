```python
# Ranking Functions (ROW_NUMBER, RANK, DENSE_RANK)

conn = sqlite3.connect(':memory:')

df_rank = pd.DataFrame({
    'department': ['IT', 'HR', 'IT', 'HR', 'HR', 'IT'],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma', 'Frank'],
    'salary': [70000, 60000, 80000, 60000, 75000, 80000]
})

# Display DataFrame
print(df_rank)
df_rank.to_sql('employees', conn, index=False, if_exists='replace')

query = """
SELECT
    department,
    name,
    salary,
    ROW_NUMBER() OVER (
        PARTITION BY department ORDER BY salary DESC
    ) AS row_num,
    RANK() OVER (
        PARTITION BY department ORDER BY salary DESC
    ) AS salary_rank,
    DENSE_RANK() OVER (
        PARTITION BY department ORDER BY salary DESC
    ) AS dense_rank
FROM employees
ORDER BY department, salary DESC;
"""

result = pd.read_sql_query(query, conn)
print('_' * 70)
print(result)
```
