## 1. Linear Space ('np.linspace')

### Overview & Purpose
Generates $N$ evenly spaced samples over a specified interval. This is particularly useful for creating smooth curves or grids for plotting and analysis.


## MATHEMATICAL FORMULATION
the step size h between consecutive points is given by:
$h = (b - a) / (N - 1)$
where $a$ and $b$ are the lower and upper bounds of the interval,
and $N$ is the number of points to generate.



```python
import matplotlib.pyplot as plt
import numpy as np
x_linear, step_size =  np.linspace(start=0.0, stop=10.0, num=15, retstep=True)
print("Linear space array: ", x_linear)
print("Step size: ", step_size)

plt.figure(figsize=(9.0,2.5))
plt.scatter(x_linear, np.zeros_like(x_linear), color='red', marker='o', s=100)

plt.vlines(x=x_linear, ymin=-0.1, ymax=0.1, color='black', linestyle='-')
plt.title("1D uniform grid Linear Space with Vertical Lines", fontsize=14)
plt.xlabel("x-axis", fontsize=12)
plt.ylabel("y-axis", fontsize=12)
plt.ylim(-0.2, 0.2)
plt.yticks([])
plt.legend(["Linear Space"])
plt.show()

```

# Logarithmic Space ('np.logspace')


```python
# Generates 7 points from 10^-5(0.00001) to 10^1(10.0) using logarithmic spacing
learning_rates = np.logspace(-5, 1, 7 , base=10)
print("Logarithmic space array: ", learning_rates)
print("Generated Hyperparameter learning rates ")
print("learning rates: ", [f"{lr:.6f}" for lr in learning_rates])

fig,(ax1,ax2) = plt.subplots(2,1,figsize=(12,4))

ax1.scatter(learning_rates, np.zeros_like(learning_rates), color='blue', marker='o', s=60, zorder=3)
ax1.set_xscale('log')
ax1.set_title("Logarithmic Space for Learning Rates", fontsize=14, fontweight='bold')
ax1.set_xlabel("Value (Log Scale)", fontsize=12)
ax1.set_yticks([])

#bottom plot : linear scale visualization
ax2.scatter(learning_rates, np.zeros_like(learning_rates), color='red', marker='o', s=100, zorder=2)
ax2.set_title("Linear Space for Learning Rates", fontsize=14, fontweight='bold')
ax2.set_xlabel("Value", fontsize=12)
ax2.set_yticks([])

plt.tight_layout()
plt.show()
```

# Geometric Space ('np.geomspace')

```python
geom_grid = np.geomspace(start=1.0, stop=1000.0, num=6)
print("Geometric space array: ", geom_grid)
print(f"Common multiplicative ratio of gemometric space: {geom_grid[1]/geom_grid[0] : .4f}")

#graphical representation of geometric space
plt.figure(figsize=(9.0,2.5))

plt.plot(range(len(geom_grid)), geom_grid, color='brown', marker='o', linestyle='-', linewidth=2, label="Geometric Space")

plt.title("Geometric Progression Growth Curve ", fontsize=14, fontweight='bold')
plt.xlabel("Iteration", fontsize=12)
plt.ylabel("Generated Value", fontsize=12)
plt.yscale('log')

plt.grid(True, which='both', ls='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
