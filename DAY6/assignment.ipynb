# Assignment 1
A temperature sensor ends conntinious readings between 20.0 degress Celsius and 25.0 degrees Celsius.
- Generate 1000 synthetic sensor readings
- compute the mean and median f your generated dataset
- generate a graphical visualization

```python
import numpy as np
import matplotlib.pyplot as plt
from tornado.autoreload import start

sensor_readings = np.random.uniform(
    low=20.0,
    high=25.0,
    size=1000
)

# print(sensor_readings)

mean_sensor_readings=np.mean(sensor_readings)
median_sensor_readings=np.median(sensor_readings)

print(mean_sensor_readings)
print(median_sensor_readings)

plt.figure(figsize=(10,5))

plt.plot(
    sensor_readings,
    alpha=0.5,
    label='sensor readings',
    linewidth=2,
    zorder=1,
    color='red'
)

plt.axhline(
    mean_sensor_readings,
    marker='x',
    markersize=10,
    label='mean',
    color='blue',
    zorder=2,
    alpha=0.5,
    linestyle='--',
    linewidth=2,
)

plt.axhline(
    median_sensor_readings,
    marker='o',
    markersize=10,
    label='median',
    color='pink',
    zorder=2,
    alpha=0.5,
    linestyle='--',
    linewidth=2,
)
plt.title('Sensor Readings')
plt.xlabel('Sensor Reading')
plt.ylabel('Temperature')
plt.legend()
plt.show()
```

# Assignment 2

E- commerce / Web Analytics (a/b testing simulator)
- task - write a script to simulate 1000 incomming website visitors
- Assign each visitor a random float value between 0.0 and 1.0 , if the value is less than 0.5, route them to design A, else route them to design B
- print the percetage of visitors routed to design A and design B

```python
import numpy as np
import matplotlib.pyplot as plt

visitors = np.random.uniform(
    low = 0.0,
    high = 1.0,
    size=1000
)

design_a = visitors[visitors < 0.5]
design_b = visitors[visitors >= 0.5]

print(f"Percentage of visitors routed to design A: {len(design_a)/len(visitors)*100:.2f}%")
print(f"Percentage of visitors routed to design B: {len(design_b)/len(visitors)*100:.2f}%")
```

# Assignemnt 3

Data Science / Computer Vision (Data Augmentation)
- Task : Imagine building an image processing pipeline for an AI model.
- Generate 500 random rotation angles for images augmentation between -15 degrees and 15 degrees.
- Plot a chart for these angles
- verify that the bar tops are roughly flat across the range rather than forming a peak in the center

```python
import numpy as np
import matplotlib.pyplot as plt

rotation_angles =  np.random.uniform(
    low=-15.0,
    high=15.0,
    size=500
)

plt.figure(figsize=(10,5))
plt.hist(rotation_angles, bins=15, alpha=0.5, label='Rotation Angles', color='red')

plt.title('Rotation Angles')
plt.xlabel('Rotation Angle')
plt.ylabel('Degrees')
plt.legend()
plt.tight_layout()
plt.show()
```

# Assignment 4
E-Commerce Delivery Time Simulation

Senario - An e-commerce platform estimates order delivery times. On average a delivery takes 30 minutes with a standard deviation of 5 minutes

tasks :
- Simulate delivery times for 1000 orders
- calculate what percentage of orders took longer than 40 minutes to deliver
- plot the delivery time and draw a threshold line at 40 minutes mark

```python
import numpy as np
import matplotlib.pyplot as plt

delivery_times = np.random.normal(loc=30, scale=5, size=1000)
print(f"Percentage of orders that took longer than 40 minutes to deliver: {len(delivery_times[delivery_times > 40])/len(delivery_times)*100:.2f}%")
plt.figure(figsize=(10,5))

plt.plot(
    delivery_times,
    alpha=0.5,
    label='Delivery Times',
    linewidth=2,
    zorder=1,
    color='red'
)

plt.axhline(
    y=40,
    color='blue',
    linestyle='--',
    linewidth=2,
    alpha=0.5,
    label='Threshold (40 minutes)'
)
plt.xlabel('Order Number')
plt.ylabel('Delivery Time (Minutes)')
plt.title('E-Commerce Delivery Time Simulation')
plt.legend()
plt.tight_layout()

plt.show()
```
