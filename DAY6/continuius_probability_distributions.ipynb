# Setup Imports and Plotting Configurations

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
from pyparsing import lineEnd

plt.style.use("seaborn-v0_8-whitegrid")


plt.rcParams['figure.dpi'] = 100
np.random.seed(42)
```

# Gaussian / Normal Distribution


```python
mu, sigma = .0, 1.5

sample_normal = np.random.normal(mu, sigma, 1000)

empirical_mean  = np.mean(sample_normal)
empirical_std = np.std(sample_normal)


print(f"Target Parameters: mu = {mu} and sigma = {sigma}")
print(f"Empirical Mean: {empirical_mean:.3f} and Empirical Standard Deviation: {empirical_std:.3f}")
print(f"Sample Range : min = {np.min(sample_normal):.3f} and max = {np.max(sample_normal):.3f}")


plt.figure(figsize=(9,3.5))
plt.style.use("seaborn-v0_8-whitegrid")
count,bins,ignored = plt.hist(sample_normal, bins=50, density=True, alpha=0.5, color='brown')
plt.xlabel("Normal Distribution")
plt.ylabel("Probability Density")
plt.title("Empirical Distribution of Normal Distribution")

pdf_normal= stats.norm.pdf(bins, loc=mu, scale=sigma)
plt.plot(bins, pdf_normal, 'r-', linewidth=2, label='PDF')
plt.legend(loc="upper right", prop={'size': 10})
plt.tight_layout()
plt.show()
```

# Continious and Uniform Distribution

## Overview & Purpose
- A continuous uniform distribution is a type of probability distribution in which all outcomes are equally likely within a specified range. It is characterized by a constant probability density function (PDF) over the interval [a, b], where a and b are the lower and upper bounds of the distribution, respectively.
- The uniform distribution is often used to represent the range of possible outcomes for a random variable, especially when there is no prior knowledge or preference for any particular value within that range. It is commonly applied in simulations, random sampling, and modeling scenarios where each outcome is equally probable.

```python
a,b = -2.0, 5.0

sample_uniform = np.random.uniform(a, b, 1000)
theoretical_mean = (a + b) / 2

theoretical_variance = ((b - a) ** 2) / 12

print(f"Interval Bounds: [ {a:.3f}, {b:.3f} ]")
print(f"Theoretical Mean: {theoretical_mean:.3f} | Empirical Mean: {np.mean(sample_uniform):.3f}")
print(f"Theoretical Variance: {theoretical_variance:.3f} | Empirical Variance: {np.var(sample_uniform):.3f}")

plt.figure(figsize=(9,3.5))
plt.hist(sample_uniform, bins=50, density=True, alpha=0.5, color='green', edgecolor='white')
x_axis = np.linspace(a-1.0, b + 1.0 , 1000)
pdf_uniform = stats.uniform.pdf(x_axis, loc=a, scale=b-a)

plt.plot(x_axis, pdf_uniform, 'r-', linewidth=2, label='PDF')
plt.title("Empirical Distribution of Uniform Distribution")
plt.xlabel("Uniform Distribution")
plt.ylabel("Probability Density")
plt.legend(loc="upper right", prop={'size': 10})
plt.tight_layout()
plt.show()
```
