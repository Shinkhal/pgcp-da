### Why EDA is important ?
- EDA is important because it helps to understand the data, identify patterns, detect anomalies, and make informed decisions based on the insights gained from the analysis. It allows data scientists and analysts to explore the data, visualize relationships, and uncover hidden insights that may not be apparent from raw data alone
- EDA can help to identify potential issues with the data, such as missing values, outliers, or inconsistencies, which can impact the accuracy of any subsequent analysis or modeling.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

df_eda = pd.DataFrame({
    'age': np.random.randint(low=18, high=70, size=200),
    'income': np.random.normal(loc=60000, scale=15000, size=200),
    'score': np.random.rand(200)
})

profile_summary = df_eda.describe().T[['mean', 'std', 'min', '25%', '50%', '75%', 'max']]
print(profile_summary)
```
