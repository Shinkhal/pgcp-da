```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha", "Riya"],
    "Age": [22, 21, np.nan, 22, 21],
    "Marks": [80, 92, 68, np.nan, 92],
    "City": ["Delhi", "Pune", "Delhi", "Mumbai", "Pune"]
})

print(df)
```

```python
print(df.isnull().sum())
print(df.isnull().sum().sum())
df["Age"] =df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```
