```python
plants = [10,20,30]

mean_val = sum(plants)/len(plants)

var_python = sum((x - mean_val) ** 2 for x in plants) / (len(plants) -1 )

var_stats = statistics.variance(plants)
std_python = math.sqrt(var_python)
std_stats = statistics.stdev(plants)
std_np = np.std(plants, ddof=1)

print(f"Python calculation : {var_python}")
print(f"Statistics module calculation : {var_stats}")
print(f"NumPy calculation : {std_np}")
print(f"Python calculation : {std_python}")
print(f"Statistics module calculation : {std_stats}")
```
