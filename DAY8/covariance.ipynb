## Covariance and Pearson Correlation Coefficient

Quick Overview:
Covariance measures how two variables change together. A positive covariance indicates that the variables tend to increase or decrease together, while a negative covariance indicates that one variable tends to increase when the other decreases.

Pearson Correlation Coefficient is a normalized version of covariance, which ranges from -1 to 1. A value of 1 indicates a perfect positive linear relationship, -1 indicates a perfect negative linear relationship, and 0 indicates no linear relationship.
